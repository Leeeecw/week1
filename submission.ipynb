# 2. Skills

Skill은 반복 작업에 필요한 지침·코드·참고자료·템플릿을 묶은 것이다.
먼저 작업을 완성한 뒤, 같은 과정을 재사용할 수 있도록 정리한다.


## 실습 1: 경제지표 수집 Skill 만들기

### 실습 목표

Yahoo Finance에서 주요 시장지표를 가져오는 Python 프로그램을 만들고, 실행 절차를 Skill로 묶는다.

**Claude 에이전트 요청 메시지**
> Yahoo Finance API를 사용하여 현재 기준으로 다음의 지표 데이터를 가져와서 정리해주는 Python 스크립트 파일을 만들어줘.
> - S&P500, 나스닥, 다우존스, 코스피, 코스닥, 미국채 10년물 금리, 금 가격, 원달러 환율

> 방금 만든 파일을 확인해서 현재의 마켓 스냅샷을 확인할 수 있는 skill을 만들어줘.
> 이 스킬을 현재 작업 디렉토리에 세팅해주고, 스킬을 ZIP 파일로도 만들어줘.

### 확인된 결과물

```text
skill-python/
└── market-snapshot/
    ├── market_summary.py
    └── market-snapshot.zip
```

### 실행

`skill-python/market-snapshot` 폴더에서 다음 명령을 실행한다.

또는 Claude Code 세션에서 다음과 같이 요청할 수 있다.
> market-snapshot Skill을 활용해서 현재 시장 상황을 표와 짧은 설명으로 정리해줘. 지표별 기준일과 수집 오류도 알려줘.

## 실습 2: 나만의 디자인으로 PPT를 만드는 Skill 만들기

### 실습 목표

디자인 규칙으로 레퍼런스 HTML 슬라이드를 만들고, 이를 활용해 PPTX를 제작한 다음 검토·수정한 작업 과정을 Skill로 패키징한다.
마지막으로 새 세션에서 Skill을 활용해 다른 발표자료를 생성한다.

**Claude 에이전트 요청 메시지**
> 첨부한 디자인 정보 텍스트를 참조해서, 레퍼런스 HTML 슬라이드 모음집을 만들어줘.
> 슬라이드라는 목표에 맞게 각 컴포넌트의 배치와 크기를 조절해줘. 폰트는 Arial을 주된 폰트로 사용해줘.

> 방금 만든 레퍼런스 HTML 슬라이드 모음집을 활용해서
> 파일 내용을 바탕으로 PPTX 슬라이드 만들어줘. 다이어그램/그래픽이 필요한 경우엔 SVG 형식으로 해줘.

> 방금 작업한 것과 같이, 레퍼런스 HTML 슬라이드 모음집을 활용해서, PPTX 슬라이드 파일을 만드는 작업을 skill로 만들어줘.
> 만든 스킬을 현재 작업 디렉토리에 세팅해주고, 스킬을 ZIP 파일로도 만들어줘.

### 확인된 결과물

```text
skill-python/
└── html-kit-to-pptx/
    ├── html-kit-to-pptx.zip
    ├── slides-reference-apple.html
    ├── DESIGN-apple.md
    ├── refer.pdf
    ├── deck-src/
    ├── deck-assets/
    ├── AI-macro-growth-deck.pptx
    └── ai-deck/
```

### 실행

Claude Code 세션에서 다음과 같이 요청할 수 있다.
> /html-kit-to-pptx Skill을 활용해서 원하는 주제로 10페이지 정도 분량의 PPTX 슬라이드를 만들어줘.
